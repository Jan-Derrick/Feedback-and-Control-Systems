```python
from openai import OpenAI
from langchain_community.chat_message_histories import ChatMessageHistory
```

```python
# 1. Configuration & Client Setup
client = OpenAI(
    base_url="http://localhost:1234/v1",
    api_key="lm-studio"
)

# Use a specific model variable for easy swapping later
MODEL_ID = "google/gemma-3-1b" 
history = ChatMessageHistory()
```

```python
print(f"--- Session Started with {MODEL_ID} ---")
print("Commands: Type 'exit', 'quit', or press Ctrl+C to stop.\n")

try:
    while True:
        # 2. User Input & Validation
        user_input = input("User (or 'exit'): ").strip()

        if user_input.lower() in ['exit', 'quit']:
            break
        
        if not user_input:
            continue

        # 3. Memory Management
        history.add_user_message(user_input)
        
        # Streamlined mapping of LangChain history to OpenAI format
        chat_context = [
            {"role": "user" if m.type == "human" else "assistant", "content": m.content}
            for m in history.messages
        ]

        # 4. Display Turn Header
        print(f"\n[YOU]: {user_input}")
        print("-" * 60)
        print("\n[Gemma 3]: ", end="", flush=True)

        # 5. Execution & Streaming
        stream = client.chat.completions.create(
            model=MODEL_ID,
            messages=chat_context,
            stream=True,
            temperature=0.7
        )

        full_ai_response = ""
        for chunk in stream:
            content = chunk.choices[0].delta.content
            if content:
                print(content, end="", flush=True)
                full_ai_response += content

        # 6. Finalize Turn
        history.add_ai_message(full_ai_response)
        print(f"\n{'=' * 60}\n")

except KeyboardInterrupt:
    print("\n\nInterrupted by user. Closing session...")
finally:
    print("Program Finished.")
```
